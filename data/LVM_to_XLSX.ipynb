```python
from pathlib import *
import warnings
import pandas as pd

warnings.filterwarnings("ignore", category=FutureWarning)
```

```python
folder = Path("/content")  # change
files = ["GAT_S1_04.11.2025.lvm", "GAT_S2_04.11.2025.lvm", "GAT_S3_12.11.2025.lvm", "GAT_S4_12.11.2025.lvm"]
sheets = ["Paraffin_54", "Beeswax_54", "Paraffin_HTPB", "Beeswax_LiAlH4"]

with pd.ExcelWriter("daq_raw_4blends.xlsx", engine="openpyxl") as writer:
    for f, sheet in zip(files, sheets):
        df = pd.read_csv(folder / f, sep="\t", skiprows=22, header=0, engine="python")
        df = df.drop(columns=[c for c in df.columns if "Comment" in str(c)], errors="ignore")
        df.to_excel(writer, sheet_name=sheet, index=False)
        print(sheet, len(df), list(df.columns))

print("Saved daq_raw_4blends.xlsx")
```
